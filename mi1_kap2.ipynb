```python
ma=[]
for j in range(100):
    x=np.random.normal(loc=10, scale=0.1, size=50)
    y=x+np.random.normal(loc=11, scale=0.2, size=50)
    g=np.corrcoef(x,y)
    ma.append(g[0,1])
ma2=np.array(ma)
ma2.std()

```
